Visualization 1: shows the relationship between the regional and conference win percentages, and the average percentages for each tourney_finish class. The result is that "more optimal" rankings like champ tend to have higher percentages for both regional and conference wins, while lower rankings like first_round_loss have lower win percentages. This is relevant to the analysis, as this visualization would indicate that certain tourney_class subcategories can be distinguished amongst each other based on these win percentages, providing the grounds for further analysis. The second scatterplot is a copy of Visualization 1, but with only the “best” and the “worst” tourney_finish subcategories.

```r
# here, I am setting up the calculation of the mean values of each tournament finishing rank, which are used in this specific visualization
tourney_rank_mean <- tidy_wnba |>
  group_by(tourney_finish) |>
  summarise(
    mean_reg = mean(reg_wins_pct, na.rm = TRUE),
    mean_conf = mean(conf_wins_pct, na.rm = TRUE)
  )
options(repr.plot.width = 15) # set up a reasonable width for my visualization, so that it is easy to see

reg_and_conf_plot_color <- tidy_wnba |>
ggplot(aes(x = reg_wins_pct, y = conf_wins_pct, color = tourney_finish)) + # choosing to make a visualization focusing on the conference win percentage agains the regional win percentage, with the different tournament finishing categories being indicated by color
  geom_point(alpha = 0.2) +   # made the raw data appear a bit lighter, so that I can more easily see the average value for each tournament finishing subcategory
  
  geom_smooth(method='lm', formula= y~x, color = "blue") + # I chose to add a regression line, so that a general directionality trend can be established
  

  geom_vline(data = tourney_rank_mean, aes(xintercept = mean_reg, color = tourney_finish), linetype = "dashed", linewidth = 1) + # horizontal line to easily indicate where the average value for each category is
  geom_hline(data = tourney_rank_mean, aes(yintercept = mean_conf, color = tourney_finish), linetype = "dashed",linewidth = 1) + # veritcal line to easily indicate where the average value for each category is
  
  geom_point(data = tourney_rank_mean, aes(x = mean_reg, y = mean_conf, color = tourney_finish),size = 5,shape = 18) + # making the average point look a little bit different, so it is possible to easily gauge its location
labs(x="Regional Win Percentage (%)", y="Conference Win Percentage(%)", color="Type of Final Tournament Ranking", title = "Distribution of Conference Win and Regional Win Percentages of Various WNBA Teams") #added human-readable labels

reg_and_conf_plot_color
```

```r

#here, I am essentially selecting the "best" and the "worst" touranment finishing categories, which are champ and first_round_loss repsectively. Hopefully by comparing the best to the worst, a trend can become apparent? Because with many different categories, it might be hard to determine.
best_and_worst_wnba <- tidy_wnba |>
filter(tourney_finish %in% c("champ", "first_round_loss"))

#here, I am basically setting up the calculation of the mean values of each tournament finishing rank, which are used in this specific visualization!
tourney_rank_mean <- best_and_worst_wnba |>
  group_by(tourney_finish) |>
  summarise(
    mean_reg = mean(reg_wins_pct, na.rm = TRUE),
    mean_conf = mean(conf_wins_pct, na.rm = TRUE)
  )

options(repr.plot.width = 15) #Set up a reasonable width for my visualization, so that it is easy to see

bestworst_reg_and_conf_plot_color <- best_and_worst_wnba |>
ggplot(aes(x = reg_wins_pct, y = conf_wins_pct, color = tourney_finish)) + #choosing to make a visualization focusing on the conference win percentage agains the regional win percentage, with the different tournament finishing categories being indicated by color
  geom_point(alpha = 0.2) +   #I made the raw data appear a bit lighter, so that I can more easily see the average value for each tournament finishing subcategory
  
  geom_smooth(method='lm', formula= y~x, color = "blue") + # I chose to add a regression line, so that a general directionality trend can be established
  

  geom_vline(data = tourney_rank_mean, aes(xintercept = mean_reg, color = tourney_finish), linetype = "dashed", linewidth = 1) + #horizontal line to easily indicate where the average value for each category is
  geom_hline(data = tourney_rank_mean, aes(yintercept = mean_conf, color = tourney_finish), linetype = "dashed",linewidth = 1) +#veritcal line to easily indicate where the average value for each category is
  
  geom_point(data = tourney_rank_mean, aes(x = mean_reg, y = mean_conf, color = tourney_finish),size = 5,shape = 18) + #making the average point look a little bit different, so it is possible to easily gauge its location
labs(x="Regional Win Percentage (%)", y="Conference Win Percentage(%)", color="Type of Final Tournament Ranking", title = "Distribution of Conference Win and Regional Win Percentages of Various WNBA Teams") #added human-readable labels!

bestworst_reg_and_conf_plot_color
```

Visualization 2: a bar plot that shows the average regional and conference win percentages, as well as the average total loss percentages, for each tournament finishing subcategory. The visualization shows that as the "better" the rank is (ie. champ), both win percentages increase, while the loss percentage decreases - the only exclusion to this trend is opening_round_loss, which has an exceptionally high avg_conference_win value. This visualization is relevant since it shows that the average values for the predictor variables demonstrate a trend accross the different tournament finishing subcategories, which could be indicative of a relationship between them and tourney_finish!

```r
options(repr.plot.width = 15) #making the visualization have a suitable length to easily view it
metric_average_info <- tidy_wnba |>
  group_by(tourney_finish) |> #lumping the data by their tournament finishing category
  summarize(avg_total_loss = mean(total_loss_pct, na.rm = TRUE),
    avg_conf_wins = mean(conf_wins_pct, na.rm = TRUE), 
       avg_reg_wins = mean(reg_wins_pct, na.rm = TRUE))|> #I am calculating the mean values for the relevant columns here (total loss percentage, conference win percentage, and regional win percentage), accounting for NA values as well.)
mutate(tourney_finish = factor(tourney_finish,levels = c("champ", "top_2_loss","top_4_loss","top_8_loss", "top_16_loss","second_round_loss", "first_round_loss", "opening_round_loss"))) #I am manually ordering all the categories from "best" to "worst", so that any trends may become apparent.

metric_average_info_long <- metric_average_info |>
  pivot_longer(cols = c(avg_total_loss, avg_conf_wins, avg_reg_wins), names_to = "type", values_to = "average") #I am using pivot longer to make my data longer --> now, there are two columns indicating the type of variable (average total loss, average conference win, and average regular wins), and the actual values themselves.

metric_average_plot <- metric_average_info_long |>
  ggplot(aes(x = tourney_finish, y = average, fill = type)) + #setting up the variables, where the x-axis represents the different tournament finishing categories, and the y value is the average percentage of the different metrics (conference win percentage, regular win percentage, and total loss percentage).
  geom_col(position = "dodge") + #I chose to put the different values side by side, since when viewed as stacked, it was hard to compare them visually. 
  labs(x = "Tournament Finishing Rank", y = "Average Percentage (%)", fill = "Percentage Type", title = "Average Regional, Conference, and Tournament Win Count based on  by Tournament Finishing Category") + #added human-readable labels!
  theme(text = element_text(size = 14.5)) +
  theme(axis.text.x = element_text(angle = 25, hjust = 1)) #the original text was sort of hard to read, so I modified the orientation and size of the text to be more space-concious.

metric_average_plot
```
